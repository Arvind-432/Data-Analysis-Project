```python
total_revenue = df['total_price'].sum()
total_pizza_sold = df['quantity'].sum()
total_orders = df['order_id'].nunique()
avg_order_value = total_revenue / total_orders
avg_pizzas_per_order = total_pizza_sold / total_orders
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Total Pizzas Sold: {total_pizza_sold:,.2f}")
print(f"Total Orders: {total_orders:,.2f}")
print(f"Average Order Value: ₹{avg_order_value:,.2f}")
print(f"Average Pizzas per Order: {avg_pizzas_per_order:,.2f}")
```

####Ingredient Analysis

```python
Ingredients =(
    df['pizza_ingredients']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .reset_index()
    .rename(columns={'index':'Count','pizza_ingredients':'Ingredients'})
)
print("Ingredients Analysis")
Ingredients.head(10)
```

####Daily Trends - Total Orders

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)
df['day_name'] = df['order_date'].dt.day_name()
weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['day_name'] = pd.Categorical(df['day_name'], categories=weekday_order, ordered=True)
orders_by_day = df.groupby('day_name', observed=False)['order_id'].nunique()
ax = orders_by_day.plot(kind='bar', figsize=(8,5),color= 'green', edgecolor='black')
plt.title('Total Orders by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_day):
  plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize = 9, fontweight = 'bold')
  plt.tight_layout()

plt.show
plt.show()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)
df['day_name'] = df['order_date'].dt.day_name()
weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['day_name'] = pd.Categorical(df['day_name'], categories=weekday_order, ordered=True)
orders_by_day = df.groupby('day_name', observed=False)['total_price'].sum()
ax = orders_by_day.plot(kind='bar', figsize=(8,5),color= 'purple', edgecolor='black')
plt.title('Total Revenue by Day of Week')
plt.xlabel('Day of the Week')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_day):
  plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize = 9, fontweight = 'bold')
  plt.tight_layout()

plt.show
plt.show()
```

####Hourly Trends - Total Orders

```python
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S')
df['order_hour'] = df['order_time'].dt.hour
orders_by_hour = df.groupby('order_hour', observed=False)['order_id'].nunique()
ax = orders_by_hour.plot(kind='bar', figsize=(10,5),color= 'orange', edgecolor='black')

plt.title('Total Orders by Hour of the Day')
plt.xlabel('Hour of the Day')
plt.ylabel('Total Orders')
plt.xticks(rotation=0)

for i, val in enumerate(orders_by_hour):
  plt.text(i, val + 5, str(val), ha='center', va='bottom', fontsize = 9, fontweight = 'bold')
  plt.tight_layout()

plt.show()
```

####Monthly Trend
****

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['month_name'] = df['order_date'].dt.month_name()

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df['month_name'] = pd.Categorical(
    df['month_name'],
    categories=month_order,
    ordered=True
)

orders_by_month = df.groupby(
    'month_name',
    observed=False
)['order_id'].nunique()

plt.figure(figsize=(10, 5))

plt.fill_between(
    range(len(orders_by_month)),
    orders_by_month.values,
    color="orange",
    alpha=0.6
)

plt.plot(
    range(len(orders_by_month)),
    orders_by_month.values,
    color="black",
    linewidth=2,
    marker='o'
)

plt.title("Total Orders by Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(
    range(len(orders_by_month)),
    orders_by_month.index,
    rotation=45
)

for i, val in enumerate(orders_by_month):
    plt.text(
        i,
        val + 20,
        str(val),
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

####Percentage of Sales Category
****

```python
category_sales = df.groupby('pizza_category')['total_price'].sum()

category_pct = category_sales / category_sales.sum() * 100

plt.figure(figsize=(7, 15))
colors = plt.get_cmap('tab20').colors  # nice color palette

plt.pie(
    category_pct,
    labels=category_pct.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'width': 0.4}
)

plt.title("Percentage of Sales by Pizza Category")
plt.show()
```

####Sales By Pizza Size and Category
****

```python
sales_pivot = df.pivot_table(
    index='pizza_category',
    columns='pizza_size',
    values='total_price',
    aggfunc='sum',
    fill_value=0
)

sales_pct = sales_pivot / sales_pivot.sum().sum() * 100

plt.figure(figsize=(10,6))
sns.heatmap(sales_pct, annot=True, fmt='.1f', cmap="YlOrRd", linewidths=0.5)
plt.title("% of Sales by Pizza Category and Size")
plt.ylabel("Pizza Category")
plt.xlabel("Pizza Size")
plt.show()
```

# Pizza Sales Analysis Overview

Exploratory data analysis (EDA) of a full year (2015) of transaction-level sales data from a pizza restaurant. Built in Google Colab using pandas, matplotlib, seaborn, and plotly.

## Dataset

- **48,620 line items** across 12 columns: order ID, pizza name/size/category, ingredients, quantity, unit/total price, order date & time
- Sourced from Google Drive as a CSV (`pizza_sales.csv`)

## What the notebook does

1. **Setup & inspection** — imports libraries, loads the data, and profiles it (`.head()`, `.info()`, `.describe()`, data types, row/column counts)
2. **Core KPIs**
3. **Ingredient analysis** — parses the multi-value `pizza_ingredients` column to find the most-used ingredients
4. **Time-based trends** — orders/revenue by day of week, hour of day, and month
5. **Category & size breakdown** — revenue share by category, plus a category × size heatmap
6. **Best/worst performers** — top 5 and bottom 5 pizzas by quantity, orders, and revenue

**Revenue by category:** Classic (26.9%), Supreme (25.5%), Chicken (24.0%), Veggie (23.7%) — fairly evenly split

**Top 5 pizzas by units sold:** The Classic Deluxe, Barbecue Chicken, Hawaiian, Pepperoni, and Thai Chicken pizzas each moved roughly 2,370–2,450 units over the year

## Purpose & style

This is a business-intelligence-style notebook — the kind used to build a restaurant sales dashboard or management report. It's descriptive, not predictive: no machine learning, statistical testing, or forecasting, just KPI reporting and visual trend analysis to answer questions like "when are we busiest?", "what sells best?", and "what should we consider dropping from the menu?"

## Notes for extending this notebook

- `order_date` and `order_time` are re-parsed from scratch in several cells rather than once up front — functional, but worth consolidating if extending the analysis
- No forecasting or ML models are included; a natural next step would be demand forecasting by day/hour or a recommendation model for menu bundling
